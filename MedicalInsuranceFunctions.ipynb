```python
def calculate_insurance_cost(name: str, age: int, sex: int, bmi: float, smoker: int, children: int):
    """
    Estimate insurance cost based on personal data.

    Parameters:
        name (str): Person's name
        age (int): Age in years
        sex (int): 0 = female, 1 = male
        bmi (float): Body Mass Index
        smoker (int): 0 = non-smoker, 1 = smoker
        children (int): Number of dependents

    Returns:
        tuple: (cost: float, message: str)
    """

    cost = (
        250 * age
        - 128 * sex
        + 370 * bmi
        + 425 * children
        + 24000 * smoker
        - 12500
    )

    message = f"{name}'s estimated insurance cost is {cost:.2f} dollars."
    return cost, message


def insurance_cost_difference(cost1: float, cost2: float):
    """
    Calculate the absolute difference between two insurance costs.

    Returns:
        tuple: (difference: float, message: str)
    """
    difference = abs(cost1 - cost2)
    message = f"The difference in insurance cost is {difference:.2f} dollars."
    return difference, message


# =========================
# Example usage
# =========================

if __name__ == "__main__":
    maria_cost, maria_msg = calculate_insurance_cost("Maria", 28, 0, 26.2, 0, 3)
    omar_cost, omar_msg = calculate_insurance_cost("Omar", 35, 1, 22.2, 0, 1)

    print(maria_msg)
    print(omar_msg)

    diff, diff_msg = insurance_cost_difference(maria_cost, omar_cost)
    print(diff_msg)
```
