```python
# =========================
# MEDICAL DATA PROCESSING
# =========================

medical_data = """Marina Allison,27,31.1,#7010.0;
Markus Valdez,30,22.4,#4050.0;
Connie Ballard,43,25.3,#12060.0;
Darnell Weber,35,20.6,#7500.0;
Sylvie Charles,22,22.1,#3022.0;
Vinay Padilla,24,26.9,#4620.0;
Meredith Santiago,51,29.3,#16330.0;
Andre Mccarty,19,22.7,#2900.0;
Lorena Hodson,65,33.1,#19370.0;
Isaac Vu,34,24.8,#7045.0"""


def clean_medical_data(data):
    # Replace '#' with '$'
    data = data.replace("#", "$")

    # Split into records
    records = data.split(";")

    cleaned_records = []

    for record in records:
        parts = record.split(",")

        if len(parts) == 4:
            name = parts[0].strip()
            age = int(parts[1].strip())
            bmi = float(parts[2].strip())
            cost = float(parts[3].replace("$", "").strip())

            cleaned_records.append({
                "name": name,
                "age": age,
                "bmi": bmi,
                "insurance_cost": cost
            })

    return cleaned_records


# =========================
# ANALYSIS FUNCTIONS
# =========================

def calculate_averages(records):
    total_bmi = sum(r["bmi"] for r in records)
    total_cost = sum(r["insurance_cost"] for r in records)

    avg_bmi = total_bmi / len(records)
    avg_cost = total_cost / len(records)

    return avg_bmi, avg_cost


# =========================
# INSURANCE MODEL
# =========================

def calculate_insurance_cost(name, age, sex, bmi, smoker, children):
    cost = (
        250 * age
        - 128 * sex
        + 370 * bmi
        + 425 * children
        + 24000 * smoker
        - 12500
    )

    message = f"{name}'s insurance cost is {cost:.2f} dollars."
    return cost, message


def insurance_cost_difference(cost1, cost2):
    difference = abs(cost1 - cost2)
    message = f"The difference in insurance cost is {difference:.2f} dollars."
    return difference, message


# =========================
# MAIN PROGRAM
# =========================

records = clean_medical_data(medical_data)

# Print cleaned names
print("Patients:")
for r in records:
    print("-", r["name"].upper())

# Calculate averages
avg_bmi, avg_cost = calculate_averages(records)
print("\nAverage BMI:", round(avg_bmi, 2))
print("Average Insurance Cost:", round(avg_cost, 2))

# Example insurance estimates
cost1, msg1 = calculate_insurance_cost("Maria", 28, 0, 26.2, 0, 3)
cost2, msg2 = calculate_insurance_cost("Omar", 35, 1, 22.2, 0, 1)

print("\n" + msg1)
print(msg2)

# Compare
diff, diff_msg = insurance_cost_difference(cost1, cost2)
print(diff_msg)
```
